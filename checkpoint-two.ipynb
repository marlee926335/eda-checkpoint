## Visualize

Create any visualizations for your EDA here. Make note in the form of code comments of what your thought process is for your visualizations.

```python
test_chart = df["instructor_id"].value_counts().head(10). #plotting the top 10 instructors by total workouts

test_chart.plot(kind='bar')
plt.title("Instructor Workout Counts")
plt.xlabel("Instructor ID")
plt.ylabel("Workout Count")
plt.show()
```

```python
test_chart_two = df["fitness_discipline_display_name"].value_counts() #plotting total number of workouts per discipline

test_chart_two.plot(kind='bar')
plt.title("Workout Counts by Fitness Discipline")
plt.xlabel("Fitness Discipline")
plt.ylabel("Workout Count")
plt.show()
```
